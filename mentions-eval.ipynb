## Speaker Attribution

```python
for novel in eval_pdnc_corefs.create_book_corefs.novels:
    eval_pdnc_corefs.create_book_corefs.booknlp_process_novel(novel)
    eval_pdnc_corefs.create_book_corefs.pdnc_process_novel(novel)
    eval_pdnc_corefs.create_book_corefs.explicit_process_novel(novel)
```

```python
qa_rows = []

for novel in eval_pdnc_corefs.create_book_corefs.novels:
    row = eval_pdnc_corefs.eval_qa_booknlp(novel)
    qa_rows.append(row)

qdf = pd.DataFrame(
    qa_rows,
    columns=['novel', 'gold_quotes', 'matched_quotes', 'pred_quotes', 'correct']
)
```

```python
qdf['quote_ident_acc'] = qdf['matched_quotes'] / qdf['gold_quotes']
qdf['attribution_acc'] = qdf['correct'] / qdf['pred_quotes']

qdf
```

```python
qdf.mean(numeric_only=True)
```
